```python
import numpy as np
import pandas as pd
import xarray as xr
import os
```

```python
# CanESM5
data_base_dir = "/Data/gfi/share/ModData/CMIP_EU_Precip_Precursors/raw/CanESM5/"
var = "pr"
var_dir = data_base_dir + var + "/"
experiment = "historical" #"ssp370" #
exp_dir = var_dir + experiment + "/"
member_no = 2
member = "r{:n}i1p1f1".format(member_no)
date_range_str_dict = {
    "historical": "18500101-20141231",
    "ssp370": "20150101-21001231"
}
file_name = var + "_day_CanESM5_" + experiment+ "_" + member + "_gn_" + date_range_str_dict[experiment] + ".nc"
file_path = exp_dir + file_name

print(file_path)
# NorESM
ref_base_dir = "/Data/gfi/share/ModData/CMIP_EU_Precip_Precursors/raw/NorESM2-LM/"
ref_var_dir = ref_base_dir + var + "/"
ref_exp_dir = ref_var_dir + experiment + "/"

ref_file_name = var + "_day_NorESM2-LM_" + experiment+ "_" + member + "_gn_" + date_range_str_dict[experiment] + ".nc"
ref_path = ref_exp_dir + ref_file_name
```

```python
file_obj = xr.open_dataset(filename_or_obj=file_path)
file_obj
```

```python
ref_obj = xr.open_dataset(filename_or_obj=ref_path)
ref_obj
```

```python
pr = file_obj.pr
pr
```

```python
pr_ref = ref_obj.pr
pr_ref
```

```python
# Select area and get the extremes for CanESM and NorESM reference
lat_Bergen = 60
lon_Bergen = 6

pr_Bergen = pr.sel(lat=lat_Bergen, lon=lon_Bergen, method="nearest") * 3600
ref_Bergen = pr_ref.sel(lat=lat_Bergen, lon=lon_Bergen, method="nearest") * 3600

pr_Bergen, ref_Bergen
```
